# App Ratings Data Analysis

This project analyzes Google Play Store app rating data to explore patterns in user ratings
and differences across app categories.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Data Loading

The dataset contains movie ratings and related metadata.

```python
df = pd.read_csv("ratings.csv")
df.head()
```

## Data Cleaning

Basic cleaning steps include checking for missing values and validating rating ranges.


```python
# Remove missing ratings
df = df[df["rating"].notna()]

# Keep only valid ratings between 1 and 5
df = df[df["rating"].between(1, 5)]
```

## Exploratory Data Analysis

Initial exploration focuses on rating distributions and category-level trends.


```python
avg_rating = df.groupby("movieId")["rating"].mean().sort_values(ascending=False)
avg_rating
```

```python
# Compute the average rating for each movie and visualize the distribution
avg_rating = df.groupby("movieId")["rating"].mean()

# Take a random sample of 1000 movies to speed up plotting
sampled_avg = avg_rating.sample(n=1000, random_state=42)

# Plot the histogram of sampled average ratings
sampled_avg.hist(bins=20)
plt.xlabel("Average Rating per Movie")
plt.ylabel("Number of Movies")
plt.title("Distribution of Average Movie Ratings (Sampled)")
plt.show()
```

## Conclusions

- The distribution of average movie ratings shows that movies tend to receive higher ratings overall.
- Differences in ratings across categories (genres) are observable, suggesting some genres are rated higher than others.
- Data cleaning (handling missing or inconsistent ratings) is essential for accurate analysis.
